## 1. Kütüphanelerin İçe Aktarılması ve Konfigürasyon

```python
# ===== KÜTÜPHANELER =====
import os
import cv2
import numpy as np
import pandas as pd
import time
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

# TensorFlow ve Keras
import tensorflow as tf
from tensorflow.keras.models import Sequential, load_model, Model
from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D, LSTM, Dense, Dropout, BatchNormalization,
    Bidirectional, GlobalAveragePooling1D, Add, Activation, LayerNormalization,
    MultiHeadAttention
)
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import TensorBoard, EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
from tensorflow.keras import regularizers

# MediaPipe
import mediapipe as mp

# Uyarıları azalt
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"

print("✓ Kütüphaneler yüklendi")
print(f"TensorFlow versiyonu: {tf.__version__}")
print(f"OpenCV versiyonu: {cv2.__version__}")
```

```python
# Konfigürasyon Parametreleri
class Config:
    # Yollar
    DATA_PATH = "MP_Data"
    MODEL_DIR = "ML_Models"
    KERAS_MODEL_PATH = os.path.join(MODEL_DIR, "action.keras")
    TFLITE_MODEL_PATH = os.path.join(MODEL_DIR, "action.tflite")
    
    # İşaret dili hareketleri (Japonca örnekler)
    ACTIONS = np.array(['konnichiwa', 'arigatou', 'gomen'])
    
    # Veri toplama parametreleri
    NO_SEQUENCES = 30  # Her hareket için kaç video
    SEQUENCE_LENGTH = 30  # Her videodaki kare sayısı
    
    # Eğitim parametreleri
    TRAIN_TEST_SPLIT = 0.2
    VALIDATION_SPLIT = 0.1
    BATCH_SIZE = 32
    LEARNING_RATE = 0.001
    MAX_EPOCHS = 1000
    EARLY_STOPPING_PATIENCE = 15
    
    # Test parametreleri
    PREDICTION_THRESHOLD = 0.4
    
    # Kamera parametreleri (Camo Studio için)
    CAMERA_INDEX = 0  # Camo Studio çalışıyor! ✅
    FRAME_WIDTH = 1280  # Camo'nun native çözünürlüğü
    FRAME_HEIGHT = 720
    
# Dizinleri oluştur
os.makedirs(Config.DATA_PATH, exist_ok=True)
os.makedirs(Config.MODEL_DIR, exist_ok=True)

print("✓ Konfigürasyon ayarlandı")
print(f"Eğitilecek hareketler: {Config.ACTIONS}")
print(f"Toplam hareket sayısı: {len(Config.ACTIONS)}")
print(f"📹 Kamera: Index {Config.CAMERA_INDEX} - Camo Studio (1280x720) ✅")
```

## 2. MediaPipe ve Yardımcı Fonksiyonlar

```python
# MediaPipe modelleri
mp_holistic = mp.solutions.holistic
mp_drawing = mp.solutions.drawing_utils
mp_drawing_styles = mp.solutions.drawing_styles

def mediapipe_detection(image, model):
    """MediaPipe modeli kullanarak görüntüde algılama yapar"""
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image.flags.writeable = False
    results = model.process(image)
    image.flags.writeable = True
    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    return image, results

def draw_landmarks(image, results):
    """Algılanan landmark'ları çizer"""
    # Yüz
    if results.face_landmarks:
        mp_drawing.draw_landmarks(
            image, results.face_landmarks, mp_holistic.FACEMESH_TESSELATION,
            landmark_drawing_spec=mp_drawing_styles.get_default_face_mesh_tesselation_style())
    
    # Poz
    if results.pose_landmarks:
        mp_drawing.draw_landmarks(
            image, results.pose_landmarks, mp_holistic.POSE_CONNECTIONS,
            landmark_drawing_spec=mp_drawing_styles.get_default_pose_landmarks_style())
    
    # Sol el
    if results.left_hand_landmarks:
        mp_drawing.draw_landmarks(
            image, results.left_hand_landmarks, mp_holistic.HAND_CONNECTIONS,
            landmark_drawing_spec=mp_drawing_styles.get_default_hand_landmarks_style())
    
    # Sağ el
    if results.right_hand_landmarks:
        mp_drawing.draw_landmarks(
            image, results.right_hand_landmarks, mp_holistic.HAND_CONNECTIONS,
            landmark_drawing_spec=mp_drawing_styles.get_default_hand_landmarks_style())

def extract_keypoints(results):
    """MediaPipe sonuçlarından anahtar noktaları çıkar"""
    pose = np.array([[res.x, res.y, res.z, res.visibility] for res in results.pose_landmarks.landmark]).flatten() if results.pose_landmarks else np.zeros(33 * 4)
    face = np.array([[res.x, res.y, res.z, res.visibility] for res in results.face_landmarks.landmark]).flatten() if results.face_landmarks else np.zeros(468 * 4)
    lh = np.array([[res.x, res.y, res.z, res.visibility] for res in results.left_hand_landmarks.landmark]).flatten() if results.left_hand_landmarks else np.zeros(21 * 4)
    rh = np.array([[res.x, res.y, res.z, res.visibility] for res in results.right_hand_landmarks.landmark]).flatten() if results.right_hand_landmarks else np.zeros(21 * 4)
    
    keypoints = np.concatenate([pose, face, lh, rh])
    
    # Sabit boyuta getir (2172)
    if len(keypoints) < 2172:
        keypoints = np.pad(keypoints, (0, 2172 - len(keypoints)))
    elif len(keypoints) > 2172:
        keypoints = keypoints[:2172]
    
    return keypoints

print("✓ MediaPipe fonksiyonları tanımlandı")
```

## 3. Veri Toplama

**Not:** Bu hücreyi çalıştırmak için webcam erişimine ihtiyacınız var. 
Veri toplama işlemi için her hareket için 30 video kaydedilecek.

**Talimatlar:**
- Kamera açıldığında, her hareket için hazırlanın
- ENTER tuşuna basarak kaydı başlatın
- Her video için hareketi yapın
- 'q' tuşuna basarak istediğiniz zaman çıkabilirsiniz

```python
def create_data_directories():
    """Veri toplama için gerekli dizinleri oluşturur"""
    for action in Config.ACTIONS:
        for sequence in range(Config.NO_SEQUENCES):
            seq_dir = os.path.join(Config.DATA_PATH, action, str(sequence))
            os.makedirs(seq_dir, exist_ok=True)
    print("✓ Veri dizinleri oluşturuldu")

create_data_directories()
```

```python
def collect_training_data():
    """Webcam kullanarak işaret dili verisi toplar"""
    print("=== Veri Toplama Başlıyor ===")
    print("Her hareket için 30 video kaydedilecek")
    print("ENTER: Kaydı başlat | Q: Çıkış\n")
    
    cap = cv2.VideoCapture(Config.CAMERA_INDEX)
    cap.set(cv2.CAP_PROP_FRAME_WIDTH, Config.FRAME_WIDTH)
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, Config.FRAME_HEIGHT)
    
    if not cap.isOpened():
        print("❌ Kamera açılamadı!")
        return
    
    with mp_holistic.Holistic(min_detection_confidence=0.5, min_tracking_confidence=0.5) as holistic:
        for action_idx, action in enumerate(Config.ACTIONS):
            print(f"\n[{action_idx+1}/{len(Config.ACTIONS)}] '{action}' hareketi için veri toplama")
            
            for sequence in range(Config.NO_SEQUENCES):
                print(f"  Video {sequence+1}/{Config.NO_SEQUENCES}")
                
                # Kullanıcı hazır olana kadar bekle
                waiting = True
                while waiting:
                    ret, frame = cap.read()
                    if not ret:
                        continue
                    
                    cv2.putText(frame, f"'{action}' - Video {sequence+1}/{Config.NO_SEQUENCES}", (15, 30),
                              cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
                    cv2.putText(frame, "ENTER: Basla | Q: Cikis", (15, 60),
                              cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
                    
                    cv2.imshow('Veri Toplama', frame)
                    
                    key = cv2.waitKey(1)
                    if key == 13:  # ENTER
                        waiting = False
                    elif key == ord('q'):
                        cap.release()
                        cv2.destroyAllWindows()
                        return
                
                # 3 saniyelik geri sayım
                for countdown in range(3, 0, -1):
                    start_time = time.time()
                    while time.time() - start_time < 1.0:
                        ret, frame = cap.read()
                        if ret:
                            cv2.putText(frame, str(countdown), (320, 240),
                                      cv2.FONT_HERSHEY_SIMPLEX, 5, (0, 0, 255), 5)
                            cv2.imshow('Veri Toplama', frame)
                            cv2.waitKey(1)
                
                # Veri toplama
                frame_count = 0
                while frame_count < Config.SEQUENCE_LENGTH:
                    ret, frame = cap.read()
                    if not ret:
                        continue
                    
                    image, results = mediapipe_detection(frame, holistic)
                    draw_landmarks(image, results)
                    
                    cv2.putText(image, "KAYIT YAPILIYOR", (20, 30),
                              cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)
                    cv2.putText(image, f"Kare: {frame_count+1}/{Config.SEQUENCE_LENGTH}", (20, 60),
                              cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)
                    
                    cv2.imshow('Veri Toplama', image)
                    
                    keypoints = extract_keypoints(results)
                    npy_path = os.path.join(Config.DATA_PATH, action, str(sequence), str(frame_count))
                    np.save(npy_path, keypoints)
                    
                    frame_count += 1
                    
                    if cv2.waitKey(1) == ord('q'):
                        cap.release()
                        cv2.destroyAllWindows()
                        return
    
    cap.release()
    cv2.destroyAllWindows()
    print("\n✓ Veri toplama tamamlandı!")

# Veri toplamak için bu satırın yorumunu kaldırın:
collect_training_data()
```

## 4. Veri Yükleme ve Hazırlama

```python
def load_dataset():
    """Toplanan veriyi yükler ve eğitim için hazırlar"""
    print("Veri yükleniyor...")
    
    label_map = {label: num for num, label in enumerate(Config.ACTIONS)}
    sequences, labels = [], []
    
    for action in Config.ACTIONS:
        for sequence in range(Config.NO_SEQUENCES):
            window = []
            for frame_num in range(Config.SEQUENCE_LENGTH):
                npy_path = os.path.join(Config.DATA_PATH, action, str(sequence), f"{frame_num}.npy")
                if os.path.exists(npy_path):
                    res = np.load(npy_path)
                    window.append(res)
            
            if len(window) == Config.SEQUENCE_LENGTH:
                sequences.append(window)
                labels.append(label_map[action])
    
    X = np.array(sequences)
    y = to_categorical(labels).astype(int)
    
    print(f"✓ Veri yüklendi: {X.shape}")
    print(f"  - Özellikler: {X.shape}")
    print(f"  - Etiketler: {y.shape}")
    
    return X, y

# Veriyi yükle
X, y = load_dataset()
```

```python
# Eğitim ve test setlerine ayır
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=Config.TRAIN_TEST_SPLIT, random_state=42)

print(f"Eğitim seti: {X_train.shape}")
print(f"Test seti: {X_test.shape}")
print(f"\nSınıf dağılımı:")
for i, action in enumerate(Config.ACTIONS):
    train_count = np.sum(y_train[:, i])
    test_count = np.sum(y_test[:, i])
    print(f"  {action}: Eğitim={train_count}, Test={test_count}")
```

## 5. Model Oluşturma

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D, GlobalAveragePooling1D,
    LSTM, Bidirectional, Dense, Dropout, BatchNormalization,
    Add, Multiply, LayerNormalization, MultiHeadAttention, Activation
)
from tensorflow.keras import regularizers

def create_model_advanced():
    """
    🚀 Gelişmiş LSTM + Attention + Residual Connections Modeli
    
    İyileştirmeler:
    ✅ Bidirectional LSTM: İleri ve geri yönlü zamansal öğrenme
    ✅ Multi-Head Attention: Önemli frame'lere odaklanma  
    ✅ Residual Connections: Derin ağda gradient akışı
    ✅ Daha derin Conv1D: Zengin uzamsal özellikler
    ✅ Layer Normalization: Eğitim stabilitesi
    ✅ Regularization: Overfitting önleme
    """
    inputs = Input(shape=(Config.SEQUENCE_LENGTH, 2172), name='input_layer')
    
    # === CONV1D BLOKLARI - Uzamsal Özellik Çıkarma ===
    
    # İlk Conv Blok
    x = Conv1D(64, kernel_size=5, padding='same', activation='relu', name='conv1d_1')(inputs)
    x = BatchNormalization()(x)
    x = Dropout(0.2)(x)
    
    # İkinci Conv Blok + Residual
    conv_res = Conv1D(128, kernel_size=1, padding='same')(x)  # Residual shortcut
    x = Conv1D(128, kernel_size=3, padding='same', activation='relu', name='conv1d_2')(x)
    x = BatchNormalization()(x)
    x = Dropout(0.2)(x)
    x = Add()([x, conv_res])  # Residual connection
    x = Activation('relu')(x)
    x = MaxPooling1D(pool_size=2)(x)
    
    # Üçüncü Conv Blok + Residual
    conv_res2 = Conv1D(256, kernel_size=1, padding='same')(x)
    x = Conv1D(256, kernel_size=3, padding='same', activation='relu', name='conv1d_3')(x)
    x = BatchNormalization()(x)
    x = Dropout(0.2)(x)
    x = Add()([x, conv_res2])
    x = Activation('relu')(x)
    
    # === BIDIRECTIONAL LSTM BLOKLARI ===
    
    # İlk Bi-LSTM katmanı
    x = Bidirectional(LSTM(128, return_sequences=True, name='lstm_1'))(x)
    x = LayerNormalization()(x)
    x = Dropout(0.3)(x)
    
    # İkinci Bi-LSTM katmanı
    lstm_out = Bidirectional(LSTM(64, return_sequences=True, name='lstm_2'))(x)
    lstm_out = LayerNormalization()(lstm_out)
    lstm_out = Dropout(0.3)(lstm_out)
    
    # === MULTI-HEAD ATTENTION MEKANİZMASI ===
    
    # Self-attention: Önemli frame'lere odaklan
    attention_out = MultiHeadAttention(
        num_heads=4,
        key_dim=64,
        dropout=0.2,
        name='multi_head_attention'
    )(lstm_out, lstm_out)
    
    # Attention + LSTM çıktılarını birleştir
    x = Add()([lstm_out, attention_out])
    x = LayerNormalization()(x)
    
    # === GLOBAL POOLING + DENSE LAYERS ===
    
    # Global Average Pooling (zamansal boyutu kaldır)
    x = GlobalAveragePooling1D()(x)
    
    # Dense katmanlar
    x = Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.001))(x)
    x = BatchNormalization()(x)
    x = Dropout(0.4)(x)
    
    x = Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.001))(x)
    x = BatchNormalization()(x)
    x = Dropout(0.4)(x)
    
    # Çıkış katmanı
    outputs = Dense(len(Config.ACTIONS), activation='softmax', name='output_layer')(x)
    
    # Modeli oluştur
    model = Model(inputs=inputs, outputs=outputs, name='SignLanguage_Advanced_Model')
    
    return model

def create_model():
    """
    Geliştirilmiş LSTM + Conv1D Hibrit Model
    (Daha basit ama yine de güçlü - önceki versiyondan çok daha iyi)
    """
    model = Sequential([
        # Daha derin Conv1D katmanları
        Conv1D(64, kernel_size=5, activation='relu', padding='same', 
               input_shape=(Config.SEQUENCE_LENGTH, 2172)),
        BatchNormalization(),
        Dropout(0.2),
        
        Conv1D(128, kernel_size=3, activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling1D(pool_size=2),
        Dropout(0.2),
        
        Conv1D(256, kernel_size=3, activation='relu', padding='same'),
        BatchNormalization(),
        MaxPooling1D(pool_size=2),
        Dropout(0.3),
        
        # Bidirectional LSTM - İleri ve geri öğrenme
        Bidirectional(LSTM(128, return_sequences=True)),
        LayerNormalization(),
        Dropout(0.3),
        
        Bidirectional(LSTM(64, return_sequences=False)),
        LayerNormalization(),
        Dropout(0.4),
        
        # Dense katmanlar
        Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
        BatchNormalization(),
        Dropout(0.4),
        
        Dense(64, activation='relu'),
        Dropout(0.3),
        
        # Çıkış
        Dense(len(Config.ACTIONS), activation='softmax')
    ], name='SignLanguage_Model')
    
    return model

# İki model seçeneği var:
# 1. create_model() - Geliştirilmiş ama basit (önerilen başlangıç için)
# 2. create_model_advanced() - En güçlü model (Attention + Residual)

# Modeli oluştur (istediğinizi seçin)
model = create_model()  # Veya create_model_advanced() kullanın
model.summary()

print("\n✅ Model oluşturuldu!")
print(f"📊 Toplam parametre: {model.count_params():,}")
print("\n💡 İpucu: Daha iyi sonuç için create_model_advanced() kullanın!")
```

## 6. Model Eğitimi

```python
# Callback'leri hazırla
log_dir = os.path.join('Logs')
os.makedirs(log_dir, exist_ok=True)

callbacks = [
    TensorBoard(log_dir=log_dir),
    EarlyStopping(
        monitor='val_loss',
        patience=Config.EARLY_STOPPING_PATIENCE,
        restore_best_weights=True,
        verbose=1
    ),
    ModelCheckpoint(
        Config.KERAS_MODEL_PATH,
        monitor='val_categorical_accuracy',
        save_best_only=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=5,
        min_lr=0.00001,
        verbose=1
    )
]

# Modeli derle
optimizer = Adam(learning_rate=Config.LEARNING_RATE)
model.compile(
    optimizer=optimizer,
    loss='categorical_crossentropy',
    metrics=['categorical_accuracy']
)

print("✓ Model derlendi")
```

```python
# Modeli eğit
print("\n=== Model Eğitimi Başlıyor ===")

history = model.fit(
    X_train, y_train,
    epochs=Config.MAX_EPOCHS,
    batch_size=Config.BATCH_SIZE,
    validation_data=(X_test, y_test),
    callbacks=callbacks,
    verbose=1
)

print("\n✓ Model eğitimi tamamlandı!")
```

## 7. Model Performansının Görselleştirilmesi

```python
# Eğitim geçmişini görselleştir
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Kayıp grafiği
axes[0].plot(history.history['loss'], label='Eğitim Kaybı')
axes[0].plot(history.history['val_loss'], label='Doğrulama Kaybı')
axes[0].set_title('Model Kaybı', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Kayıp')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Doğruluk grafiği
axes[1].plot(history.history['categorical_accuracy'], label='Eğitim Doğruluğu')
axes[1].plot(history.history['val_categorical_accuracy'], label='Doğrulama Doğruluğu')
axes[1].set_title('Model Doğruluğu', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Doğruluk')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 8. Model Değerlendirmesi

```python
# Test setinde değerlendir
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nTest Kaybı: {test_loss:.4f}")
print(f"Test Doğruluğu: {test_acc:.4f} ({test_acc*100:.2f}%)")
```

```python
# Confusion Matrix
y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = np.argmax(y_test, axis=1)

cm = confusion_matrix(y_true_classes, y_pred_classes)

# Confusion matrix'i görselleştir
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=Config.ACTIONS, 
            yticklabels=Config.ACTIONS)
plt.title('Confusion Matrix', fontsize=16, fontweight='bold')
plt.ylabel('Gerçek Etiket')
plt.xlabel('Tahmin Edilen Etiket')
plt.tight_layout()
plt.show()

# Sınıflandırma raporu
print("\nSınıflandırma Raporu:")
print(classification_report(y_true_classes, y_pred_classes, target_names=Config.ACTIONS))
```

```python
# Her sınıf için doğruluk oranları
class_accuracies = []
for i, action in enumerate(Config.ACTIONS):
    mask = y_true_classes == i
    if np.sum(mask) > 0:
        acc = np.sum((y_pred_classes == i) & mask) / np.sum(mask)
        class_accuracies.append(acc)
    else:
        class_accuracies.append(0)

# Sınıf bazlı doğruluk grafiği
plt.figure(figsize=(12, 6))
bars = plt.bar(Config.ACTIONS, class_accuracies, color='steelblue', alpha=0.8)
plt.axhline(y=test_acc, color='r', linestyle='--', label=f'Ortalama: {test_acc:.2%}')
plt.title('Sınıf Bazında Doğruluk Oranları', fontsize=14, fontweight='bold')
plt.xlabel('Hareket')
plt.ylabel('Doğruluk')
plt.ylim(0, 1.1)
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.grid(True, alpha=0.3, axis='y')

# Bar üzerine değerleri yaz
for bar, acc in zip(bars, class_accuracies):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{acc:.1%}',
             ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

## 9. Modeli Kaydetme

```python
# Keras modeli zaten ModelCheckpoint ile kaydedildi
print(f"✓ Keras modeli kaydedildi: {Config.KERAS_MODEL_PATH}")

# TFLite modeline dönüştür
try:
    converter = tf.lite.TFLiteConverter.from_keras_model(model)
    converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS, tf.lite.OpsSet.SELECT_TF_OPS]
    converter._experimental_lower_tensor_list_ops = False
    tflite_model = converter.convert()
    
    with open(Config.TFLITE_MODEL_PATH, 'wb') as f:
        f.write(tflite_model)
    
    print(f"✓ TFLite modeli kaydedildi: {Config.TFLITE_MODEL_PATH}")
except Exception as e:
    print(f"⚠ TFLite dönüşümü başarısız: {e}")
```

## 10. Test Tahminleri

```python
# Rastgele test örnekleri seç ve tahmin yap
n_samples = 5
random_indices = np.random.choice(len(X_test), n_samples, replace=False)

print("\nÖrnek Tahminler:")
print("=" * 60)

for idx in random_indices:
    sample = X_test[idx:idx+1]
    true_label = Config.ACTIONS[np.argmax(y_test[idx])]
    
    prediction = model.predict(sample, verbose=0)[0]
    predicted_label = Config.ACTIONS[np.argmax(prediction)]
    confidence = np.max(prediction)
    
    status = "✓" if true_label == predicted_label else "✗"
    print(f"{status} Gerçek: {true_label:12s} | Tahmin: {predicted_label:12s} | Güven: {confidence:.2%}")
    
    # Tüm sınıflar için olasılıkları göster
    print("  Olasılık dağılımı:")
    for action, prob in zip(Config.ACTIONS, prediction):
        bar = '█' * int(prob * 30)
        print(f"    {action:12s}: {bar} {prob:.2%}")
    print()
```
