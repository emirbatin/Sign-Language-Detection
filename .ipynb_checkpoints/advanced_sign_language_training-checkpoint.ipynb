## 1. Kütüphaneler ve Konfigürasyon

```python
# Temel kütüphaneler
import os
import cv2
import numpy as np
import pandas as pd
import time
from pathlib import Path

# Görselleştirme
import matplotlib.pyplot as plt
import seaborn as sns

# ML kütüphaneleri
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

# TensorFlow ve Keras
import tensorflow as tf
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import (
    Input, Dense, Dropout, LSTM, Bidirectional, 
    Concatenate, BatchNormalization, LayerNormalization,
    Conv1D, MaxPooling1D, GlobalAveragePooling1D,
    MultiHeadAttention, Add, Activation
)
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import (
    TensorBoard, EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras import regularizers

# MediaPipe
import mediapipe as mp

# GPU ayarları
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print(f"✅ GPU bulundu: {len(gpus)} adet")
        print(f"   {gpus[0].name}")
    except RuntimeError as e:
        print(e)
else:
    print("⚠️ GPU bulunamadı, CPU kullanılacak")

print(f"\n📦 TensorFlow: {tf.__version__}")
print(f"📦 OpenCV: {cv2.__version__}")
print("✅ Tüm kütüphaneler yüklendi")
```

```python
# Konfigürasyon
class Config:
    # Veri yolları
    BASE_DIR = Path("Advanced_Data")
    LANDMARKS_DIR = BASE_DIR / "landmarks"
    OPTICAL_FLOW_DIR = BASE_DIR / "optical_flow"
    MODEL_DIR = Path("Advanced_Models")
    
    # Model yolları
    KERAS_MODEL_PATH = MODEL_DIR / "sign_language_advanced.keras"
    TFLITE_MODEL_PATH = MODEL_DIR / "sign_language_advanced.tflite"
    
    # İşaret dili hareketleri
    ACTIONS = np.array([
        'konnichiwa', 'arigatou', 'gomen'
    ])
    
    # Veri toplama parametreleri
    NO_SEQUENCES = 50  # Her hareket için video sayısı (30'dan fazla = daha iyi)
    SEQUENCE_LENGTH = 30  # Her videodaki frame sayısı
    
    # Özellik boyutları
    LANDMARK_FEATURES = 2172  # MediaPipe landmarks
    FLOW_HEIGHT = 64  # Optical flow yükseklik (küçültülmüş)
    FLOW_WIDTH = 64   # Optical flow genişlik (küçültülmüş)
    FLOW_CHANNELS = 2  # Optical flow kanalları (dx, dy)
    
    # Eğitim parametreleri
    TRAIN_TEST_SPLIT = 0.2
    BATCH_SIZE = 16  # GPU için optimize
    LEARNING_RATE = 0.0001
    MAX_EPOCHS = 200
    EARLY_STOPPING_PATIENCE = 25
    
    # Test parametreleri
    PREDICTION_THRESHOLD = 0.5
    
    # Kamera parametreleri
    CAMERA_INDEX = 0  # Camo Studio için
    FRAME_WIDTH = 1280
    FRAME_HEIGHT = 720

# Dizinleri oluştur
for action in Config.ACTIONS:
    (Config.LANDMARKS_DIR / action).mkdir(parents=True, exist_ok=True)
    (Config.OPTICAL_FLOW_DIR / action).mkdir(parents=True, exist_ok=True)
Config.MODEL_DIR.mkdir(exist_ok=True)

print("✅ Konfigürasyon ayarlandı")
print(f"📊 Hareketler: {Config.ACTIONS}")
print(f"📊 Toplam: {len(Config.ACTIONS)} hareket")
print(f"📊 Her hareket için {Config.NO_SEQUENCES} video toplanacak")
print(f"📊 Toplam veri: {len(Config.ACTIONS) * Config.NO_SEQUENCES} video")
```

## 2. Feature Extraction Fonksiyonları

```python
# MediaPipe setup
mp_holistic = mp.solutions.holistic
mp_drawing = mp.solutions.drawing_utils
mp_drawing_styles = mp.solutions.drawing_styles

def mediapipe_detection(image, model):
    """MediaPipe ile landmark tespiti"""
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image.flags.writeable = False
    results = model.process(image)
    image.flags.writeable = True
    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
    return image, results

def draw_landmarks(image, results):
    """Landmark'ları görselleştir"""
    if results.pose_landmarks:
        mp_drawing.draw_landmarks(
            image, results.pose_landmarks, mp_holistic.POSE_CONNECTIONS,
            landmark_drawing_spec=mp_drawing_styles.get_default_pose_landmarks_style())
    
    if results.left_hand_landmarks:
        mp_drawing.draw_landmarks(
            image, results.left_hand_landmarks, mp_holistic.HAND_CONNECTIONS,
            landmark_drawing_spec=mp_drawing_styles.get_default_hand_landmarks_style())
    
    if results.right_hand_landmarks:
        mp_drawing.draw_landmarks(
            image, results.right_hand_landmarks, mp_holistic.HAND_CONNECTIONS,
            landmark_drawing_spec=mp_drawing_styles.get_default_hand_landmarks_style())

def extract_keypoints(results):
    """MediaPipe landmarks'tan özellik çıkar"""
    pose = np.array([[res.x, res.y, res.z, res.visibility] 
                     for res in results.pose_landmarks.landmark]).flatten() \
           if results.pose_landmarks else np.zeros(33*4)
    
    face = np.array([[res.x, res.y, res.z] 
                     for res in results.face_landmarks.landmark]).flatten() \
           if results.face_landmarks else np.zeros(468*3)
    
    lh = np.array([[res.x, res.y, res.z] 
                   for res in results.left_hand_landmarks.landmark]).flatten() \
         if results.left_hand_landmarks else np.zeros(21*3)
    
    rh = np.array([[res.x, res.y, res.z] 
                   for res in results.right_hand_landmarks.landmark]).flatten() \
         if results.right_hand_landmarks else np.zeros(21*3)
    
    keypoints = np.concatenate([pose, face, lh, rh])
    
    # Sabit boyut: 2172
    if len(keypoints) < Config.LANDMARK_FEATURES:
        keypoints = np.pad(keypoints, (0, Config.LANDMARK_FEATURES - len(keypoints)))
    elif len(keypoints) > Config.LANDMARK_FEATURES:
        keypoints = keypoints[:Config.LANDMARK_FEATURES]
    
    return keypoints

print("✅ MediaPipe fonksiyonları hazır")
```

```python
# Optical Flow fonksiyonları

def calculate_optical_flow(prev_gray, curr_gray):
    """
    Farneback Optical Flow hesapla
    Returns: (flow_x, flow_y) - hareket vektörleri
    """
    flow = cv2.calcOpticalFlowFarneback(
        prev_gray, curr_gray, 
        None,
        pyr_scale=0.5,
        levels=3,
        winsize=15,
        iterations=3,
        poly_n=5,
        poly_sigma=1.2,
        flags=0
    )
    
    return flow

def resize_optical_flow(flow, target_size=(64, 64)):
    """
    Optical flow'u küçült (hafızayı azaltmak için)
    """
    flow_resized = cv2.resize(flow, target_size, interpolation=cv2.INTER_LINEAR)
    return flow_resized

def visualize_optical_flow(flow):
    """
    Optical flow'u HSV renk uzayında görselleştir
    """
    h, w = flow.shape[:2]
    
    # Magnitude ve angle hesapla
    magnitude, angle = cv2.cartToPolar(flow[..., 0], flow[..., 1])
    
    # HSV görselleştirme
    hsv = np.zeros((h, w, 3), dtype=np.uint8)
    hsv[..., 0] = angle * 180 / np.pi / 2  # Hue = yön
    hsv[..., 1] = 255  # Saturation = max
    hsv[..., 2] = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX)  # Value = büyüklük
    
    # BGR'e çevir
    bgr = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)
    return bgr

print("✅ Optical Flow fonksiyonları hazır")
```

## 3. Veri Toplama

**Bu sistem şunları kaydeder:**
1. MediaPipe landmarks (2172 özellik)
2. Optical Flow (64x64x2 - hareket vektörleri)

**Kontroller:**
- ENTER: Kaydı başlat
- Q: Çıkış

```python
def collect_advanced_data():
    """
    Gelişmiş veri toplama: Landmarks + Optical Flow
    """
    print("="*60)
    print("🎬 GELİŞMİŞ VERİ TOPLAMA BAŞLIYOR")
    print("="*60)
    print(f"📊 {len(Config.ACTIONS)} hareket için {Config.NO_SEQUENCES} video toplanacak")
    print(f"📊 Toplam: {len(Config.ACTIONS) * Config.NO_SEQUENCES} video")
    print(f"\n⌨️ Kontroller:")
    print(f"   ENTER: Kaydı başlat")
    print(f"   Q: Çıkış")
    print("="*60 + "\n")
    
    cap = cv2.VideoCapture(Config.CAMERA_INDEX)
    cap.set(cv2.CAP_PROP_FRAME_WIDTH, Config.FRAME_WIDTH)
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, Config.FRAME_HEIGHT)
    
    if not cap.isOpened():
        print("❌ Kamera açılamadı!")
        return
    
    with mp_holistic.Holistic(
        min_detection_confidence=0.5,
        min_tracking_confidence=0.5
    ) as holistic:
        
        for action_idx, action in enumerate(Config.ACTIONS):
            print(f"\n{'='*60}")
            print(f"🎯 [{action_idx+1}/{len(Config.ACTIONS)}] '{action}' hareketi")
            print(f"{'='*60}")
            
            for sequence in range(Config.NO_SEQUENCES):
                print(f"\n  📹 Video {sequence+1}/{Config.NO_SEQUENCES}", end="")
                
                # Kullanıcı hazır olana kadar bekle
                waiting = True
                while waiting:
                    ret, frame = cap.read()
                    if not ret:
                        continue
                    
                    # Bilgi yazıları
                    cv2.putText(frame, f"Hareket: {action}", (15, 30),
                               cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
                    cv2.putText(frame, f"Video: {sequence+1}/{Config.NO_SEQUENCES}", (15, 70),
                               cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
                    cv2.putText(frame, "ENTER: Basla | Q: Cikis", (15, 110),
                               cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2)
                    
                    cv2.imshow('Gelismis Veri Toplama', frame)
                    
                    key = cv2.waitKey(1)
                    if key == 13:  # ENTER
                        waiting = False
                    elif key == ord('q'):
                        cap.release()
                        cv2.destroyAllWindows()
                        return
                
                # 3-2-1 geri sayım
                for countdown in range(3, 0, -1):
                    start_time = time.time()
                    while time.time() - start_time < 1.0:
                        ret, frame = cap.read()
                        if ret:
                            cv2.putText(frame, str(countdown), (300, 300),
                                      cv2.FONT_HERSHEY_SIMPLEX, 5, (0, 0, 255), 10)
                            cv2.imshow('Gelismis Veri Toplama', frame)
                            cv2.waitKey(1)
                
                # Veri toplama
                landmarks_sequence = []
                flow_sequence = []
                prev_gray = None
                
                for frame_num in range(Config.SEQUENCE_LENGTH):
                    ret, frame = cap.read()
                    if not ret:
                        continue
                    
                    # MediaPipe detection
                    image, results = mediapipe_detection(frame, holistic)
                    draw_landmarks(image, results)
                    
                    # Landmarks çıkar
                    keypoints = extract_keypoints(results)
                    landmarks_sequence.append(keypoints)
                    
                    # Optical Flow hesapla
                    curr_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
                    if prev_gray is not None:
                        flow = calculate_optical_flow(prev_gray, curr_gray)
                        flow_resized = resize_optical_flow(flow, 
                                                          (Config.FLOW_WIDTH, Config.FLOW_HEIGHT))
                        flow_sequence.append(flow_resized)
                        
                        # Optical flow görselleştir (küçük bir köşede)
                        flow_vis = visualize_optical_flow(flow)
                        flow_vis_small = cv2.resize(flow_vis, (160, 120))
                        image[10:130, image.shape[1]-170:image.shape[1]-10] = flow_vis_small
                    else:
                        # İlk frame - sıfır flow
                        flow_sequence.append(np.zeros((Config.FLOW_HEIGHT, 
                                                      Config.FLOW_WIDTH, 2)))
                    
                    prev_gray = curr_gray
                    
                    # Bilgi yazıları
                    cv2.putText(image, "KAYIT YAPILIYOR", (20, 30),
                               cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)
                    cv2.putText(image, f"Frame: {frame_num+1}/{Config.SEQUENCE_LENGTH}", (20, 60),
                               cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
                    cv2.putText(image, "Optical Flow", (image.shape[1]-160, 140),
                               cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255, 255, 0), 1)
                    
                    cv2.imshow('Gelismis Veri Toplama', image)
                    
                    if cv2.waitKey(1) == ord('q'):
                        cap.release()
                        cv2.destroyAllWindows()
                        return
                
                # Verileri kaydet
                landmarks_array = np.array(landmarks_sequence)
                flow_array = np.array(flow_sequence)
                
                np.save(Config.LANDMARKS_DIR / action / f"{sequence}.npy", landmarks_array)
                np.save(Config.OPTICAL_FLOW_DIR / action / f"{sequence}.npy", flow_array)
                
                print(" ✅ Kaydedildi")
    
    cap.release()
    cv2.destroyAllWindows()
    
    print("\n" + "="*60)
    print("✅ VERİ TOPLAMA TAMAMLANDI!")
    print("="*60)

# Veri toplamak için çalıştırın:
collect_advanced_data()
```

## 4. Veri Yükleme ve Hazırlama

```python
def load_advanced_dataset():
    """
    Landmarks + Optical Flow verilerini yükle
    Returns: (X_landmarks, X_flow, y)
    """
    print("📂 Veri yükleniyor...")
    
    label_map = {label: num for num, label in enumerate(Config.ACTIONS)}
    
    landmarks_list = []
    flow_list = []
    labels_list = []
    
    for action in Config.ACTIONS:
        for sequence in range(Config.NO_SEQUENCES):
            # Landmarks yükle
            landmarks_path = Config.LANDMARKS_DIR / action / f"{sequence}.npy"
            flow_path = Config.OPTICAL_FLOW_DIR / action / f"{sequence}.npy"
            
            if landmarks_path.exists() and flow_path.exists():
                landmarks = np.load(landmarks_path)
                flow = np.load(flow_path)
                
                # Boyut kontrolü
                if landmarks.shape[0] == Config.SEQUENCE_LENGTH and \
                   flow.shape[0] == Config.SEQUENCE_LENGTH:
                    landmarks_list.append(landmarks)
                    flow_list.append(flow)
                    labels_list.append(label_map[action])
    
    # Array'lere çevir
    X_landmarks = np.array(landmarks_list)
    X_flow = np.array(flow_list)
    y = to_categorical(labels_list, num_classes=len(Config.ACTIONS)).astype(int)
    
    print(f"✅ Veri yüklendi!")
    print(f"   Landmarks: {X_landmarks.shape}")
    print(f"   Optical Flow: {X_flow.shape}")
    print(f"   Labels: {y.shape}")
    print(f"   Toplam örnek: {len(y)}\")\n")
    
    return X_landmarks, X_flow, y

# Veriyi yükle
X_landmarks, X_flow, y = load_advanced_dataset()
```

```python
# Train/Test split
X_landmarks_train, X_landmarks_test, X_flow_train, X_flow_test, y_train, y_test = train_test_split(
    X_landmarks, X_flow, y, 
    test_size=Config.TRAIN_TEST_SPLIT, 
    random_state=42,
    stratify=np.argmax(y, axis=1)  # Balanced split
)

print(f"📊 Veri setleri:")
print(f"   Eğitim: {len(y_train)} örnek")
print(f"   Test: {len(y_test)} örnek")
print(f"\n📊 Sınıf dağılımı:")
for i, action in enumerate(Config.ACTIONS):
    train_count = np.sum(y_train[:, i])
    test_count = np.sum(y_test[:, i])
    print(f"   {action:15s}: Train={train_count:3.0f}, Test={test_count:3.0f}")
```

## 5. Dual-Stream Model Mimarisi

**İki Dal (Stream):**
1. **Landmarks Stream:** MediaPipe noktalarını işler (Bi-LSTM)
2. **Optical Flow Stream:** Hareket vektörlerini işler (Conv3D + LSTM)
3. **Fusion:** İki stream birleştirilir ve final tahmin

**Avantajlar:**
- Landmarks titrese bile Flow kurtarır
- Flow yeterli değilse Landmarks kurtarır
- %90-95 doğruluk hedefi

```python
from tensorflow.keras.layers import Conv3D, MaxPooling3D

def create_dual_stream_model():
    """
    🔥 Dual-Stream Architecture
    
    Stream 1: MediaPipe Landmarks → Bi-LSTM
    Stream 2: Optical Flow → Conv3D → LSTM
    Fusion → Final prediction
    """
    
    # ===== STREAM 1: LANDMARKS (MediaPipe) =====
    landmarks_input = Input(
        shape=(Config.SEQUENCE_LENGTH, Config.LANDMARK_FEATURES), 
        name='landmarks_input'
    )
    
    # Bi-LSTM katmanları
    x1 = Bidirectional(LSTM(128, return_sequences=True))(landmarks_input)
    x1 = LayerNormalization()(x1)
    x1 = Dropout(0.3)(x1)
    
    x1 = Bidirectional(LSTM(64, return_sequences=False))(x1)
    x1 = LayerNormalization()(x1)
    x1 = Dropout(0.4)(x1)
    
    x1 = Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.001))(x1)
    x1 = BatchNormalization()(x1)
    x1 = Dropout(0.4)(x1)
    
    # ===== STREAM 2: OPTICAL FLOW =====
    flow_input = Input(
        shape=(Config.SEQUENCE_LENGTH, Config.FLOW_HEIGHT, Config.FLOW_WIDTH, Config.FLOW_CHANNELS),
        name='flow_input'
    )
    
    # 3D Convolutions (spatial + temporal)
    x2 = Conv3D(32, kernel_size=(3, 3, 3), activation='relu', padding='same')(flow_input)
    x2 = BatchNormalization()(x2)
    x2 = MaxPooling3D(pool_size=(2, 2, 2))(x2)
    x2 = Dropout(0.2)(x2)
    
    x2 = Conv3D(64, kernel_size=(3, 3, 3), activation='relu', padding='same')(x2)
    x2 = BatchNormalization()(x2)
    x2 = MaxPooling3D(pool_size=(2, 2, 2))(x2)
    x2 = Dropout(0.3)(x2)
    
    # Reshape for LSTM: (batch, time, features)
    x2 = tf.keras.layers.Reshape(
        (-1, x2.shape[2] * x2.shape[3] * x2.shape[4])
    )(x2)
    
    # Temporal modeling
    x2 = LSTM(64, return_sequences=False)(x2)
    x2 = LayerNormalization()(x2)
    x2 = Dropout(0.4)(x2)
    
    x2 = Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.001))(x2)
    x2 = BatchNormalization()(x2)
    x2 = Dropout(0.4)(x2)
    
    # ===== FUSION: İki stream'i birleştir =====
    merged = Concatenate()([x1, x2])
    
    # Final layers
    merged = Dense(256, activation='relu', kernel_regularizer=regularizers.l2(0.001))(merged)
    merged = BatchNormalization()(merged)
    merged = Dropout(0.5)(merged)
    
    merged = Dense(128, activation='relu')(merged)
    merged = Dropout(0.4)(merged)
    
    # Output
    outputs = Dense(len(Config.ACTIONS), activation='softmax', name='output')(merged)
    
    # Model oluştur
    model = Model(
        inputs=[landmarks_input, flow_input],
        outputs=outputs,
        name='DualStream_SignLanguage'
    )
    
    return model

# Modeli oluştur
model = create_dual_stream_model()
model.summary()

print(f"\n✅ Model oluşturuldu!")
print(f"📊 Toplam parametre: {model.count_params():,}")
print(f"📊 Stream 1: Landmarks (Bi-LSTM)")
print(f"📊 Stream 2: Optical Flow (Conv3D + LSTM)")
print(f"📊 Fusion: Concatenate + Dense layers")
```

## 6. Model Eğitimi (RTX 5060)

```python
# Callbacks
log_dir = Path('Advanced_Logs')
log_dir.mkdir(exist_ok=True)

callbacks = [
    TensorBoard(log_dir=str(log_dir)),
    
    EarlyStopping(
        monitor='val_loss',
        patience=Config.EARLY_STOPPING_PATIENCE,
        restore_best_weights=True,
        verbose=1
    ),
    
    ModelCheckpoint(
        str(Config.KERAS_MODEL_PATH),
        monitor='val_categorical_accuracy',
        save_best_only=True,
        verbose=1
    ),
    
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=10,
        min_lr=0.000001,
        verbose=1
    )
]

# Model compile
optimizer = Adam(learning_rate=Config.LEARNING_RATE)
model.compile(
    optimizer=optimizer,
    loss='categorical_crossentropy',
    metrics=['categorical_accuracy']
)

print("✅ Model derlendi!")
print(f"   Optimizer: Adam (lr={Config.LEARNING_RATE})")
print(f"   Loss: Categorical Crossentropy")
print(f"   Metrics: Categorical Accuracy")
```

```python
# Model eğitimi
print("\n" + "="*60)
print("🚀 MODEL EĞİTİMİ BAŞLIYOR (RTX 5060)")
print("="*60)
print(f"📊 Batch Size: {Config.BATCH_SIZE}")
print(f"📊 Max Epochs: {Config.MAX_EPOCHS}")
print(f"📊 Early Stopping Patience: {Config.EARLY_STOPPING_PATIENCE}")
print("="*60 + "\n")

history = model.fit(
    [X_landmarks_train, X_flow_train],  # İki input
    y_train,
    epochs=Config.MAX_EPOCHS,
    batch_size=Config.BATCH_SIZE,
    validation_data=([X_landmarks_test, X_flow_test], y_test),
    callbacks=callbacks,
    verbose=1
)

print("\n✅ EĞİTİM TAMAMLANDI!")
```

## 7. Model Değerlendirmesi

```python
# Eğitim grafiği
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Loss
axes[0].plot(history.history['loss'], label='Eğitim', linewidth=2)
axes[0].plot(history.history['val_loss'], label='Doğrulama', linewidth=2)
axes[0].set_title('Model Kaybı', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Accuracy
axes[1].plot(history.history['categorical_accuracy'], label='Eğitim', linewidth=2)
axes[1].plot(history.history['val_categorical_accuracy'], label='Doğrulama', linewidth=2)
axes[1].set_title('Model Doğruluğu', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('training_history.png', dpi=150, bbox_inches='tight')
plt.show()

print("✅ Eğitim grafiği kaydedildi: training_history.png")
```

```python
# Test seti değerlendirmesi
test_loss, test_acc = model.evaluate(
    [X_landmarks_test, X_flow_test], 
    y_test, 
    verbose=0
)

print("\n" + "="*60)
print("📊 TEST SETİ SONUÇLARI")
print("="*60)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f} ({test_acc*100:.2f}%)")
print("="*60)

# Tahminler
y_pred = model.predict([X_landmarks_test, X_flow_test], verbose=0)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = np.argmax(y_test, axis=1)

# Confusion Matrix
cm = confusion_matrix(y_true_classes, y_pred_classes)

plt.figure(figsize=(12, 10))
sns.heatmap(cm, annot=True, fmt='d', cmap='YlGnBu',
            xticklabels=Config.ACTIONS,
            yticklabels=Config.ACTIONS,
            cbar_kws={'label': 'Count'})
plt.title('Confusion Matrix - Dual Stream Model', fontsize=16, fontweight='bold', pad=20)
plt.ylabel('Gerçek Etiket', fontsize=12)
plt.xlabel('Tahmin Edilen Etiket', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()

# Classification Report
print("\n" + "="*60)
print("📊 SINIFLANDIRMA RAPORU")
print("="*60)
print(classification_report(y_true_classes, y_pred_classes, target_names=Config.ACTIONS))
print("="*60)
```

## 8. Model Kaydetme
